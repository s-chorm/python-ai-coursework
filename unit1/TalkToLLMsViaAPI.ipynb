## Make the First Model API Call

In the cell below, you're going to make your very first API call using OpenAI's **Chat Completion API**. This is one of the main API endpoints used for interacting with models like GPT-4o, and it’s designed specifically for handling conversational exchanges.

There are two key elements I would like you to notice:
1. The `model` is specified explicitly. As model providers release more options, telling the API exactly which one you want becomes essential.
2. The `messages` structure. It's a list containing a dictionary with `"role"` and `"content"`. This is how we provide our input text, known as the prompt, which is the instruction or question for the LLM.

#### Run the Following Cell

````python
response = client.chat.completions.create(
        model="gpt-4o", # Model name needs to be specified explicitly 
        messages=[{"role": "user", "content": "Hello AI"}] # The prompt is the text ("Hello AI"), assigned to "content" 
    )

display(Markdown(f"{response.choices[0].message.content}"))
````

Congratulations! You just made your first API call to a model, the first of many more to come! You can see how the model responded specifically to our prompt *"Hello AI"*. The API returns a structured response that contains a list called `"choices"`. Even when the model produces only one reply, the API still wraps it inside this list.

Visually, it looks like this:
```text
response
└── choices (a list of completion choices)
    └── [0] (the first choice)
        └── message (the model's message object, with "role" and "content")
            └── content (the model's actual text response)
```

The code `response.choices[0].message.content` we used to extract the response simply navigates through this structure to reach the model’s answer.

Now let's look at some of the metadata that comes along with the response. Notice how we use similar dot (`.`) notation to access different parts of the Chat Completions API response object.

#### Run the Following Cell

````python
md_content = f"""
**Response Details**
- **Model:** `{response.model}`
- **Tokens used:** `{response.usage.total_tokens}`
"""

display(Markdown(md_content))
````

You will see that the returned fields include the specific model version, and the number of tokens used in the conversation so far, both input and output. Note that we have displayed just two of the many fields returned in the API response. Many of these are helpful metadata fields that are used for observability and debugging purposes. 

💡 **Key insight:**  Remember that tokens are smaller parts of a word, and hence the token count is generally higher than the total number of words that you will see in a conversation. A helpful rule of thumb is that one token generally corresponds to ~4 characters of text for common English text.

## Make the API Call Structure Reusable

Since we will be making many API calls, it can be helpful to wrap this logic in a function. This keeps our code clean, avoids repetition, and ensures we send prompts and extract responses in a consistent way. Note that wrapping the logic in a function is generally optional, because you could write the code directly each time, but using a function can make the codebase much easier to manage. The specifics will depend on the software or API you are working with, but the goal here is to show you the underlying pattern so you can recognize and apply it in any context when needed.

The cell below defines the function called `query_llm`.

#### Run the Following Cell

````python
def query_llm(prompt: str, model: str = "gpt-4o") -> str:
    """
    Sends a user-provided prompt to LLM model and returns the response.

    Parameters:
    prompt (str): The text input that will be sent to the language model.

    Returns:
    str: The model's generated response based on the given prompt.

    Example:
    query_llm("What is the capital of France?")
    "The capital of France is Paris."

    """
    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}]
    )
    return response.choices[0].message.content
````

### Write a Prompt and Call the Function

Let’s explore what happens when we ask the model to identify itself and describe its capabilities. The variable `prompt_1` contains the prompt we will send to the model. As you will observe in the function above, the default model we will use is `gpt-4o`. Before running the cell below, can you try to guess the expected response? Find out by running the code cell below.

#### Run the Following Cell

````python
prompt_1 = "Hi, which LLM model are you and what are your capabilities?"

display(Markdown(query_llm(prompt_1)))
````

Interesting, isn't it, as to how the model *knows itself*? Later we will see whether we can introduce oursleves to the model, and see if it remembers our name.
